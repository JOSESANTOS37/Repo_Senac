Utilizando a base de

#1.Filtrar os dados relacionados a furtos de celulares.

#2.Agrupar os dados por delegacia, somando o total
de furtos registrados.

#3.Aplicar análise estatística descritiva:
•Calcular a média, mediana e os quartis das
ocorrências por delegacia.

#4.Identificar:
•As delegacias que representam os 25% com
menor número de furtos (primeiro quartil).
•As delegacias que representam os 25% com
maior número de furtos de celular.

```python
ispdados = pd.read_csv('https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv', sep=';',encoding='latin1')
ispdados
```

```python

# 1. Filtrar os dados relacionados a furtos de celulares
furtos_celulares = ispdados.sort_values(by ="furto_celular")
furtos_celulares


```

```python
# 2. Agrupar os dados por delegacia (cisp), somando os furtos de celular

delegacias = furtos_celulares.groupby('cisp')['furto_celular'].sum().reset_index()
delegacias



```

```python
#3.Aplicar análise estatística descritiva:•Calcular a média, mediana e os quartis dasocorrências por delegacia.

array_furtos_celular = np.array(delegacias["furto_celular"])

# Mediana (valor central)
mediana = np.median(array_furtos_celular)

# Quartis
q1  = np.percentile(array_furtos_celular, 25)  # Primeiro quartil
q2  = np.percentile(array_furtos_celular, 50)  # Segundo quartil (igual à mediana)
q3 = np.percentile(array_furtos_celular, 75)  # Terceiro quartil

print(f'Primeiro quartil (Q1): {q1}')
print(f'Segundo quartil  (Q2, Mediana): {q2}')
print(f'Mediana {mediana}')
print(f'Terceio quartil  (Q3): {q3}')


```

```python
#4.Identificar:•As delegacias que representam os 25% com menor número de furtos (primeiro quartil).
#•As delegacias que representam os 25% com maior número de furtos de celular.

array_furtos_celular = np.array(delegacias["furto_celular"])

# Mediana (valor central)
mediana = np.median(array_furtos_celular)


# Quartis
q1  = np.percentile(array_furtos_celular, 25)  # Primeiro quartil
q3 = np.percentile(array_furtos_celular, 75)  # Terceiro quartil

print(f'Primeiro quartil (Q1): {q1}')
print(f'Terceio quartil  (Q3): {q3}')
```

    Desafio analítico 02

O Secretário de Segurança Pública do Estado solicitou uma análise acerca das ocorrências estelionato registradas por
município. O objetivo é identificar àqueles municípios que representam os 25% que mais ocorreram tal fato. Além
disso, o Secretário quer identificar aqueles que menos ocorreram.

```python
# Agrupar os dados por município (munic), somando os estelionatos

estelionato_municipio = ispdados.groupby('munic')['estelionato'].sum().reset_index()
estelionato_municipio

```

```python
array_estelinato = np.array(estelionato_municipio["estelionato"])

# Mediana (valor central)
mediana = np.median(array_estelinato)


# Quartis
q1  = np.percentile(array_estelinato, 25)  # Primeiro quartil
q3 = np.percentile(array_estelinato, 75)  # Terceiro quartil

print(f'Primeiro quartil (Q1): {q1}')
print(f'Terceio quartil  (Q3): {q3}')
```
