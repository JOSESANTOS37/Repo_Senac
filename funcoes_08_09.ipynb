```python
## Atividade 1

def ler_par_e_somar(numero_par):
    print(f'Soma dos {numero_par} pares:')
    a = int(input('Digite o 1 número: '))
    b = int(input('Digite o 2 número: '))
    soma = a + b
    return soma

# Executa a função para 3 pares
for i in range(1, 4):
    resultado = ler_par_e_somar(i)
    print(f"Soma do par {i}: {resultado}")


```

```python
## Atividade 2

def multa(velocidade, passada):
    excesso = ((passada - velocidade) / velocidade) * 100

    if passada <= velocidade:
        print("Sem multa.")
    elif excesso <= 10:
        print(f"Seu veículo excedeu em {excesso:.2f}% e sua multa leve é de R$ 180,00.")
    elif excesso <= 30:
        print(f"Seu veículo excedeu em {excesso:.2f}% e sua multa média é de R$ 340,56.")
    else:
        print(f"Seu veículo excedeu em {excesso:.2f}% e sua multa grave é de R$ 480,98.")
```

```python
multa(100,131)
```

```python
## Atividade 3

def taxa_metabolica(sexo, peso, altura, idade):
    if sexo == 'm':  # masculino
        tmb = 66.5 + (13.75 * peso) + (5.003 * altura) - (6.755 * idade)
        print(f"Sua taxa metabólica basal é {tmb:.2f} kcal")

    elif sexo == 'f':  # feminino
        tmb = 655.1 + (9.563 * peso) + (1.850 * altura) - (4.676 * idade)
        print(f"Sua taxa metabólica basal é {tmb:.2f} kcal")
         
```

```python
taxa_metabolica('f',80,182,27)
```
