```python
# Importa a biblioteca pandas, que é usada para manipulação e análise de dados
import pandas as pd

# Importa a biblioteca matplotlib.pyplot, que é usada para criar gráficos
import matplotlib.pyplot as plt

# Cria um DataFrame (tabela) com duas colunas: "Ano" e "Vendas"
df = pd.DataFrame({
    'Ano':    [2020, 2021, 2022, 2023],
    'Vendas': [100, 150, 130, 180]})

df['Ano'] = df['Ano'].  astype(str)

# Cria um gráfico de linha (line plot) usando os dados do DataFrame
# No eixo X: os anos
# No eixo Y: as vendas correspondentes
plt.plot(df['Ano'], df['Vendas'])

# Define o título do gráfico
plt.title('Vendas por Ano')

# Define o rótulo do eixo X
plt.xlabel('Ano')

# Define o rótulo do eixo Y
plt.ylabel('Vendas')

# Adiciona uma grade de fundo (linhas horizontais e verticais para facilitar a leitura)
plt.grid(True)

# Exibe o gráfico na tela
plt.show()
```

```python
# Cria um DataFrame com duas colunas: "Produto" e "Vendas"
df = pd.DataFrame({
    'Produto': ['A', 'B', 'C'],
    'Vendas':  [120, 90, 150]})

# Ordena o DataFrame pela coluna "Vendas" em ordem decrescente
df.sort_values(by= 'Vendas', ascending= False, inplace= True) #inplace=True → significa que a alteração será feita no próprio DataFrame df, sem precisar criar uma cópia.

# Cria um gráfico de barras
# - Eixo X: nomes dos produtos
# - Eixo Y: número de vendas
plt.bar(df['Produto'], df['Vendas'])

# Define o título do gráfico
plt.title('Vendas por Produto')

# Define o rótulo do eixo X
plt.xlabel('Produto')

# Define o rótulo do eixo Y
plt.ylabel('Vendas')

# Exibe o gráfico na tela
plt.show()
```

```python
# Cria um dicionário com duas colunas: "Ano" e "Vendas"
dados = {
    'Ano':    [2018, 2019, 2020, 2021, 2022],
    'Vendas': [150, 200, 230, 300, 330]
}

# Converte o dicionário em um DataFrame
df = pd.DataFrame(dados)

# Define o tamanho da figura (8 de largura x 5 de altura)
plt.figure(figsize=(8, 5))

# Cria um gráfico de barras horizontais
# - eixo Y: anos
# - eixo X: vendas
# - cor: preto
plt.barh(df['Ano'], df['Vendas'], color='blue')

# Define o título do gráfico
plt.title('Vendas por Ano')

# Define o rótulo do eixo X
plt.xlabel('Vendas')

# Define o rótulo do eixo Y
plt.ylabel('Ano')

# Salva o gráfico como arquivo PNG
plt.savefig('grafico_barras_horizontais.png')

# Exibe o gráfico na tela
plt.show()
```

```python

# Cria um dicionário com os dias da semana e as vendas correspondentes
dados = {
    'Dias':   ['Segunda', 'Terça', 'Quarta', 'Quinta', 'Sexta'],
    'Vendas': [500, 200, 1000, 150, 1950]
}

# Converte o dicionário em DataFrame para facilitar a manipulação
df = pd.DataFrame(dados)

# Extrai as colunas para variáveis auxiliares
dias = df['Dias']
valores = df['Vendas']

# Cria um gráfico de linha
# marker='o' → adiciona bolinhas em cada ponto
plt.plot(dias, valores, marker='o', linestyle='-', color='blue')

# Define o título do gráfico
plt.title('Vendas por Dias')

# Define os rótulos dos eixos
plt.xlabel('Dia da semana')
plt.ylabel('Valor Vendido')

# Exibe o gráfico
plt.show()
```
