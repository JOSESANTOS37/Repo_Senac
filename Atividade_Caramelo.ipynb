```python
import pandas as pd
```

```python
caramelo = pd.read_csv('CARAMELO.csv', sep= ';')
caramelo
```

```python
# vendas agrupadas pelo estado do Rio de Janeiro

df_Estado = caramelo.loc[caramelo['Cidade Loja'] == 'Rio de Janeiro']
df_Estado 
```

```python
# Tratamento de conversão de string numérica para float

caramelo['Valor de Venda'] = caramelo ['Valor de Venda'].str.replace(',','.').astype(float)
```

```python
# vendas pelo método de pagamento 

df_Metodo_Pagamento = caramelo.groupby('Metodo de Pagamento')['Valor de Venda'].sum().reset_index()
df_Metodo_Pagamento
```

```python
# Vendas pelo de sexo 

df_Metodo_Pagamento = caramelo.groupby('Sexo')['Valor de Venda'].sum().reset_index()
df_Metodo_Pagamento
```

```python
# Vendas por lojas 

df_Metodo_Pagamento = caramelo.groupby('Loja')['Valor de Venda'].sum().reset_index()
df_Metodo_Pagamento
```
