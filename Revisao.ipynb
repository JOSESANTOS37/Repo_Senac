8. Listas e Decisão

- Crie uma lista com idades.
- Use for e if para exibir quem é maiorde idade.

```python

idades = [12, 17, 18, 20, 15, 30]


for idade in idades:
  
    if idade >= 18:
        print(f"Idade {idade}: maior de idade")
    else:
        print(f"Idade {idade}: menor de idade")
```

9. Cadastro Simples

- Crie uma lista vazia.
- Peça 3 nomes ao usuário e adicione à lista.
- Exiba a lista final.

```python
nomes = []

# Pede 3 nomes ao usuário e adiciona na lista
for i in range(3):
    nome = input(f"Digite o {i+1}º nome: ")
    nomes.append(nome)  # adiciona cada nome à lista

# Exibe a lista final
print("Lista final de nomes:", nomes)
```
