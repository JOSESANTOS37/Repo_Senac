```python
# importa a biblioteca pandas e dá o apelido 'pd'
import pandas as pd  

# importa a biblioteca numpy e dá o apelido 'np'
import numpy as np 
```

```python
# Importa os dados do arquivo 

df = pd.read_csv('https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv',delimiter=";", encoding="latin1")

# Exibe o conteúdo completo do DataFrame 'pesquisa' no console.
df.head()
```

```python
municipios_baixada_fluminense = [
   "Duque de Caxias",
    "Nova Iguaçu",
    "São João de Meriti",
    "Belford Roxo",
    "Nilópolis",
    "Mesquita",
    "Queimados",
    "Japeri",
    "Magé",
    "Guapimirim",
    "Seropédica",
    "Itaguaí",
    "Paracambi"
]
```

```python




municipios = df.loc[(df['munic'].isin(municipios_baixada_fluminense))]
municipios
```

```python
df_baixada = municipios.groupby(['munic'])['aaapai'].sum().reset_index()
df_baixada

```

```python
q3 = np.percentile(df_baixada["aaapai"], 75)
print(f'Terceiro quartil (Q3) = {q3}')
```

```python
df_final = df_baixada.loc[df_baixada["aaapai"] > q3]
df_final
```

```python
df_final.to_csv('Apreensões_menores.csv', sep=";", index=  False)
```

```python
array_df_baixada = np.array(df_baixada["aaapai"])

# Mediana (valor central)
mediana_anterior = np.median(array_df_baixada)
media_anterior = np.mean(array_df_baixada)

# Quartis
q1_anterior = np.percentile(array_df_baixada, 25)  # Primeiro quartil
q2_anterior = np.percentile(array_df_baixada, 50)  # Terceiro quartil
q3_anterior = np.percentile(array_df_baixada, 75)  # Terceiro quartil
distancia_anterior = (media_anterior - mediana_anterior) / mediana_anterior
iqr_anterior = q3_anterior - q1_anterior
limite_superior_total = q3_anterior + (1.5*iqr_anterior)
limite_inferior_total = q1_anterior - (1.5*iqr_anterior)

print(f'media {media_anterior}')
print(f'mediana {mediana_anterior}')
print(f'distancia {distancia_anterior*100}%')
```

```python
dados_array = np.array(array_df_baixada)

media = np.mean(dados_array)

variancia = np.var(dados_array)

desvio_padrao = np.std(dados_array)

coeficiente_variacao = (desvio_padrao/ media)*100

distancia_variancia = (variancia / (media **2))*100

print("Média:", media)

print("Variância:", variancia)

print("Desvio padrão:", desvio_padrao)

print("Coeficiente de variação:", coeficiente_variacao)

print("Distância da variância em relação à média:", distancia_variancia)
```

```python

Outliers_superior = df_final.loc[df_final['aaapai'] >= limite_superior_total]
print(Outliers_superior)
```

```python
Outliers_superior.to_csv('Outliers_superior _Apreensões.csv', sep=";", index=  False)
```
