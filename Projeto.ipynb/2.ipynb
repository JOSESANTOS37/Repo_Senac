```python
# Importa os dados do arquivo 

df = pd.read_csv('https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv',delimiter=";", encoding="latin1")

# Exibe o conteúdo completo do DataFrame 'pesquisa' no console.
df.head()
```

```python
Rio_de_janeiro = [
   "Rio de Janeiro"
]
```

```python

municipio = df.loc[(df['munic'].isin(Rio_de_janeiro))]
municipio
```

```python
df_municipio = municipio.groupby(['cisp'])['estelionato'].sum().reset_index()
df_municipio

```

```python
q3 = np.percentile(df_municipio["estelionato"], 75)
print(f'Terceiro quartil (Q3) = {q3}')
```

```python
df_final = df_municipio.loc[df_municipio["estelionato"] > q3]
df_final
```

```python
df_final.to_csv('Rio de janeiro_estelionato.csv', sep=";", index=  False)
```

```python
array_df_municipio = np.array(df_municipio["estelionato"])

# Mediana (valor central)
mediana_anterior = np.median(array_df_municipio)
media_anterior = np.mean(array_df_municipio)

# Quartis
q1_anterior = np.percentile(array_df_municipio, 25)  # Primeiro quartil
q2_anterior = np.percentile(array_df_municipio, 50)  # Terceiro quartil
q3_anterior = np.percentile(array_df_municipio, 75)  # Terceiro quartil
distancia_anterior = (media_anterior - mediana_anterior) / mediana_anterior
iqr_anterior = q3_anterior - q1_anterior
limite_superior_total = q3_anterior + (1.5*iqr_anterior)
limite_inferior_total = q1_anterior - (1.5*iqr_anterior)

print(f'media {media_anterior}')
print(f'mediana {mediana_anterior}')
print(f'distancia {distancia_anterior*100}%')
```

```python

Outliers_superior = df_final.loc[df_final['estelionato'] >= limite_superior_total]
print(Outliers_superior)
```

```python
Outliers_superior.to_csv('Outliers_superior_Estelionatos.csv', sep=";", index=  False)
```
