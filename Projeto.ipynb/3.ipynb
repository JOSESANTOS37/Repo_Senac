```python
# importa a biblioteca pandas e dá o apelido 'pd'
import pandas as pd  

# importa a biblioteca numpy e dá o apelido 'np'
import numpy as np 
```

```python
# Importa os dados do arquivo 

df = pd.read_csv('https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv',delimiter=";", encoding="latin1")

# Exibe o conteúdo completo do DataFrame 'pesquisa' no console.
df.head()
```

```python
municipios_regiao_dos_lagos = [
    "Araruama",
    "Armação dos Búzios",
    "Arraial do Cabo",
    "Cabo Frio",
    "Iguaba Grande",
    "São Pedro da Aldeia",
    "Saquarema"
]

```

```python

regiao_lagos = df.loc[(df['munic'].isin(municipios_regiao_dos_lagos))]
regiao_lagos
```

```python
violencia = [
    "hom_doloso",
    "lesao_corp_morte",
    "latrocinio",
    "cvli",
    "hom_por_interv_policial",
    "letalidade_violenta",
    "tentat_hom",
    "lesao_corp_dolosa",
    "estupro"
]
```

```python
crimes_lagos = regiao_lagos[["munic","hom_doloso",
    "lesao_corp_morte",
    "latrocinio",
    "cvli",
    "hom_por_interv_policial",
    "letalidade_violenta",
    "tentat_hom",
    "lesao_corp_dolosa",
    "estupro",]] 
crimes_lagos

```

```python
crimes_lagos["crimes_violentos"] = crimes_lagos["hom_doloso"].astype(int) + crimes_lagos["lesao_corp_morte"].astype(int) + crimes_lagos["latrocinio"].astype(int) + crimes_lagos["cvli"].astype(int) + crimes_lagos["hom_por_interv_policial"].astype(int) + crimes_lagos["letalidade_violenta"].astype(int) + crimes_lagos["tentat_hom"].astype(int) + crimes_lagos["lesao_corp_dolosa"].astype(int) + crimes_lagos["estupro"].astype(int)
crimes_lagos
```
